````python
# Check data structure
print(log_df.info())  # to confirm 65532 entries, 12 columns, data types

# Convert port columns to categorical (object type) since they are categorical codes
# Use the renamed column names here
port_cols = ["source_port", "destination_port", "nat_source_port", "nat_destination_port"]
for col in port_cols:
    log_df[col] = log_df[col].astype(str)  # treat ports as strings (categorical)

# Handle categorical levels by grouping rare categories into 'Other'
# Define a helper to group infrequent categories below a threshold frequency
def group_infrequent(df, col, threshold):
    freq = df[col].value_counts(normalize=True)
    frequent_vals = freq[freq >= threshold].index
    df[col] = df[col].where(df[col].isin(frequent_vals), "Other")
    return df

# Apply grouping on each port column with chosen thresholds (determined from EDA)
# Use the renamed column names here:
log_df = group_infrequent(log_df, "source_port", threshold=0.00021)       # ~ top 4 classes kept&#8203;:contentReference[oaicite:55]{index=55}
log_df = group_infrequent(log_df, "nat_source_port", threshold=0.00004)   # ~ top 2 classes kept&#8203;:contentReference[oaicite:56]{index=56}
log_df = group_infrequent(log_df, "destination_port", threshold=0.004)    # ~ top 5 classes kept&#8203;:contentReference[oaicite:57]{index=57}
log_df = group_infrequent(log_df, "nat_destination_port", threshold=0.1)  # ~ top 5 classes kept&#8203;:contentReference[oaicite:58]{index=58}

# One-hot encode the categorical port columns
log_cat = pd.get_dummies(log_df[port_cols], drop_first=False)
print(log_cat.shape)  # examine number of dummy columns for ports

# Scale numerical features
# Updated num_cols list to match the actual column names in log_df
num_cols = ["bytes", "bytes_sent", "bytes_receieved", "packets", "elapsed_time"]
scaler = StandardScaler()
log_num_scaled = pd.DataFrame(scaler.fit_transform(log_df[num_cols]), columns=num_cols)

# Combine processed numeric and dummy features
log_features = pd.concat([log_num_scaled, log_cat], axis=1)
log_target = log_df["Action"]  # target variable

print("Final feature matrix shape:", log_features.shape)  # should show high dimensional feature set
print(log_features.columns[:10])  # sample of feature columns for verification
````

````python

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC

# Split data into train and test sets (stratify by target to preserve class distribution)
X_train, X_test, y_train, y_test = train_test_split(log_features, log_target, test_size=0.2,
                                                    stratify=log_target, random_state=0)

# Define an SVM model and hyperparameter grid for linear kernel
svm_model = SVC(kernel="linear")
param_grid = {"C": [0.0001, 0.001, 0.01, 0.1, 0.5, 1.0]}  # range of regularization strengths

grid_search = GridSearchCV(svm_model, param_grid, cv=3, scoring="accuracy", verbose=1)
grid_search.fit(X_train, y_train)
print("Best C for linear SVM:", grid_search.best_params_)

# Train the best linear SVM on full training data
best_linear_svm = grid_search.best_estimator_
best_linear_svm.fit(X_train, y_train)

# (Optional) Train and evaluate a polynomial SVM for comparison
poly_svm = SVC(kernel="poly", degree=3, C=grid_search.best_params_["C"], cache_size=2000)
poly_svm.fit(X_train, y_train)
````

### VW Performance Evaluation Results:
The Vowpal Wabbit models were trained and evaluated across different values of L2 regularization. The key findings included:

- Models performed exceptionally well overall (~99% accuracy), especially on the predominant classes (**allow, deny, drop**).
- VW outperformed SVM models in classification accuracy and efficiency, particularly notable with lighter computational demands.
- The VW model, however, encountered challenges with accurately predicting the rare "reset-both" class, similarly to the other tested algorithms (SVM and SGD).

**VW Hyperparameter Tuning Results Summary (Accuracy)**:
- L2 = 1: Slightly lower accuracy, indicating over-regularization.
- L2 = 0.1: Improved accuracy, still suboptimal.
- L2 = 0.001 and L2 = 0.0001: Achieved highest accuracy levels.
- Further decreases in regularization (below 0.00001) provided no significant gains, indicating diminishing returns.

The **optimal hyperparameter value** selected was **L2 = 0.0001**, offering a robust balance between accuracy and generalization, as observed in the evaluation metrics.
